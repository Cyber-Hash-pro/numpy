Arithematic operation

```python
a1=np.array([11,22,33])
a2=np.array([44,55,66])
a1+a2 
a1 * a2
a1 / a2 
a1 //  a2 
#[ a1[0]+a2[0],a1[1]+a2[1],a1[2]+a2[2]] only when size same
```

| Operator | Meaning                       | Example        |
| -------- | ----------------------------- | -------------- |
| `+`      | Element-wise Addition         | `11+44 = 55`   |
| `-`      | Element-wise Subtraction      | `11-44 = -33`  |
| `*`      | Element-wise Multiplication   | `11*44 = 484`  |
| `/`      | True Division (float)         | `11/44 = 0.25` |
| `//`     | Floor Division (integer part) | `11//44 = 0`   |
| `%`      | Remainder                     | `11%44 = 11`   |
| `**`     | Power                         | `11**2 = 121`  |
